```python
import numpy as np
import pandas as pd
from scipy import stats

# Set random seed for reproducibility
np.random.seed(42)

# 0. CARILAH DATA SEBANYAK 100 TAHAP/SAMPEL
# Generating 100 random samples (e.g., normally distributed data with mean = 50, std = 10)
n = 100
data = np.random.normal(loc=50.0, scale=10.0, size=n)

print(f"Jumlah Sampel (n): {len(data)}")
print(f"Data Pertama (80 sampel awal): {data[:100]}\n")

# 1. A. ESTIMASI TITIK DAN INTERVAL (Z-DISTRIBUTION, 5% ERROR)
# Significance level (alpha) = 5% = 0.05
alpha = 0.05

# Point estimation (Sample Mean)
point_estimate = np.mean(data)

# Standard deviation of sample and standard error
sample_std = np.std(data, ddof=1)
standard_error = sample_std / np.sqrt(n)

# Z value for a two-tailed test with alpha = 0.05 (Z_{0.025} = 1.96)
z_score = stats.norm.ppf(1 - alpha / 2)

# Interval estimation (Confidence Interval)
margin_of_error = z_score * standard_error
ci_lower = point_estimate - margin_of_error
ci_upper = point_estimate + margin_of_error

print("--- A. ESTIMASI TITIK & INTERVAL (Distribusi Z, Kesalahan 5%) ---")
print(f"Nilai Estimasi Titik (Rata-rata Sampel): {point_estimate:.4f}")
print(f"Z-score (1-alpha/2): {z_score:.4f}")
print(f"Interval Kepercayaan 95%: [{ci_lower:.4f}, {ci_upper:.4f}]\n")

# 1. B. RESAMPLING: JACKKNIFE & BOOTSTRAPPING
# --- 1. Jackknife Method ---
# Jackknife leaves out one observation at a time
jackknife_means = []
for i in range(n):
    resample = np.delete(data, i)
    jackknife_means.append(np.mean(resample))

jackknife_mean_estimate = np.mean(jackknife_means)

print("--- B1. JACKKNIFE RESAMPLING ---")
print(f"Jumlah Iterasi Jackknife: {len(jackknife_means)}")
print(f"Rata-rata dari Hasil Resampling Jackknife: {jackknife_mean_estimate:.4f}\n")

# --- 2. Bootstrapping Method (1000 iterations) ---
n_iterations = 1000
bootstrap_means = []

for _ in range(n_iterations):
    # Resample with replacement
    resample = np.random.choice(data, size=n, replace=True)
    bootstrap_means.append(np.mean(resample))

bootstrap_mean_estimate = np.mean(bootstrap_means)

print("--- B2. BOOTSTRAPPING RESAMPLING (1000 Iterasi) ---")
print(f"Jumlah Iterasi Bootstrapping: {n_iterations}")
print(f"Rata-rata dari Hasil Resampling Bootstrapping: {bootstrap_mean_estimate:.4f}")
```
